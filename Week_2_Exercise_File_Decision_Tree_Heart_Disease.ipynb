# Heart Disease#

The data contain observations for patients with potential heart problems. The task is to predict whether a patient has a heart disease based on their profile and a series of medical tests.

**Variables**

**AGE**	Numerical	Age (years)

**SEX**	Binary	Gender

**CHESTPAIN**	Categorical	The type of chest-pain experienced:

0 = typical angina

1 = atypical angina

2 = non -anginal pain

3 = asymptotic

**RESTBP**	Numerical	Resting blood pressure (beats per minute)

**CHOL**	Numerical	Serum cholesterol (mg/dl)

**SUGAR**	Binary	The fasting blood sugar value of an individual with 120mg/dl.

If fasting blood sugar > 120mg/dl then : 1 (true)

else : 0 (false)

**ECG**	Categorical	3-valued resting electrocardiographic results:

0 = normal

1 = having ST-T wave abnormality

2 = left ventricular hyperthrophy

**MAXHR**	Numerical	Maximum heart rate achieved (beats per minute)

**ANGINA**	Binary	Was angina induced by exercise?

1 = yes

0 = no

**DEP**	Numerical	ST depression induced by exercise relative to rest

**EXERCISE**	Ordered categorical	Slope of the peak exercise ST segment

0 = upsloping

1 = flat

2 = downsloping

**FLUOR**	Numerical	The number of major vessels colored by fluoroscopy

**THAL**	Categorical	3-valued Thal:

1 = normal

2 = fixed defect

3 = reversible defect

**Target**	Binary	Presence or absence of a heart disease

1: Presence

0: Absence
